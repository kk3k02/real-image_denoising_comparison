# Denoising Results Analysis

Loads evaluation CSV files, builds aggregate tables, and creates figures for Chapter 5.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

results_dir = Path('../results')
csvs = [p for p in results_dir.glob('*.csv') if p.name not in {'summary.csv'} and not p.name.startswith('benchmark_')]
frames = []
for path in csvs:
    method, domain, dataset = path.stem.split('_', 2)
    frame = pd.read_csv(path)
    frame['method'] = method
    frame['domain'] = domain
    frame['dataset'] = dataset
    frames.append(frame)
results = pd.concat(frames, ignore_index=True) if frames else pd.DataFrame()
results.head()
```

```python
summary = pd.read_csv(results_dir / 'summary.csv') if (results_dir / 'summary.csv').exists() else pd.DataFrame()
summary
```

```python
if not results.empty:
    ax = results.groupby(['dataset', 'domain', 'method'])['psnr'].mean().unstack('method').plot(kind='bar', figsize=(10, 4))
    ax.set_ylabel('PSNR [dB]')
    ax.set_title('Mean PSNR by dataset/domain')
    plt.tight_layout()
    plt.show()
```

```python
benchmark_csvs = list(results_dir.glob('benchmark_*.csv'))
benchmarks = pd.concat([pd.read_csv(p) for p in benchmark_csvs], ignore_index=True) if benchmark_csvs else pd.DataFrame()
benchmarks
```
